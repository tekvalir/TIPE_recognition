# Requirements

```python
import os
import matplotlib.pyplot as plt 
import numpy as np
import cv2
import uuid
```

# Image Resizer

```python
for file in os.listdir("data\\images"):
    if file.endswith(".jpg"):
        image = plt.imread(os.path.join("data\\images\\", file))
        res = cv2.resize(image, dsize=(800,450), interpolation=cv2.INTER_CUBIC)
        plt.imsave(os.path.join("data\\images\\", file), res)
```

# Mask creator

```python
img = plt.imread(os.path.join('data', 'images', '18c31b86-b2d6-11ed-9fa7-0242ac1c000c.jpg'))
```

```python
test = plt.imread(os.path.join('data', 'labels', '18c31b86-b2d6-11ed-9fa7-0242ac1c000c_json', 'label.png'))
```

```python
for i in range(0,450):
    for j in range(0,800):
        if test[i,j,0] != 0: 
            test[i,j] = [1,1,1,1]
            
A=np.zeros((test.shape[0], test.shape[1]), dtype=int)
for i in range(0,450):
    for j in range(0,800):
        if test[i,j,0] != 0: 
            A[i,j]=1
```

```python
fig, ax = plt.subplots(ncols=2, figsize=(20,20))
ax[0].imshow(test)
ax[1].imshow(A, cmap='gray')
plt.show()
```

```python
plt.imsave(os.path.join('data', 'masks', '18c31b86-b2d6-11ed-9fa7-0242ac1c000c.png'), A, cmap='gray') # Save image
```
