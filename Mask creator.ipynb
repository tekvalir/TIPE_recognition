# Mask creator

## Requirements and Settings

```python
import os
import numpy as np
import matplotlib.pyplot as plt
import json
import cv2
import sys
from PIL import Image, ImageDraw
```

```python
plt.rcParams["figure.figsize"]=10,10
```

## Data Processing

```python
def data_resizer(fp,height,width):
    for file in os.listdir(os.path.join(fp, 'images')):
        if file.endswith(".jpg") or file.endswith('.png'):
            image = Image.open(os.path.join(fp, 'images', file))
            image = image.resize((width,height))
            image.save(os.path.join(fp, 'images', file))
            
def data_mask_creator(fp,height,width):    
    for file in os.listdir(os.path.join(fp, 'labels')):
        if file.endswith(".json"):
            with open(os.path.join('data', 'labels', file)) as content:
                data = np.round(np.array(json.load(content)['shapes'][0]['points'])).astype(int)
            mask = Image.new("1", (width,height))
            draw = ImageDraw.Draw(mask)
            for i in range (data.shape[0]):
                draw.line([(data[i,0], data[i,1]), (data[(i+1) % data.shape[0],0], data[(i+1) % data.shape[0],1])], fill=1)
            xb,yb=1/data.shape[0]*np.sum(data[:,0]),1/data.shape[0]*np.sum(data[:,1])
            ImageDraw.floodfill(mask, (xb,yb), 1)
            mask.save((os.path.join(fp, 'masks', ''.join(file.split('.')[:-1]+['.png']))))
```

```python
data_mask_creator('data',450,800)
```
