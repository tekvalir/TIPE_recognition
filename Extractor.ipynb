# Extraction d'images depuis une vidéo

```python
import cv2
import os
import uuid
import numpy as np
from PIL import Image
```

```python
def extract_features(video_input_file_path, feature_output_file_path, lapse):
    count = 0
    print('Extracting frames from video: ', video_input_file_path)
    vidcap = cv2.VideoCapture(video_input_file_path)
    success, image = vidcap.read()
    features = []
    success = True
    while success:
        vidcap.set(cv2.CAP_PROP_POS_MSEC, (count * lapse * 1000))
        success, image = vidcap.read()
        if success:
            cv2.imwrite(os.path.join(feature_output_file_path, str(uuid.uuid1()) + ".jpg"), image)
            count += 1
    return print(str(count) + " images extracted")
```

Extraction depuis une vidéo enregistrée de chemin _video\_input\_file\_path_ d'images toutes les _lapse_ secondes et stockées dans le dossier de chemin _feature\_output\_file\_path_.

```python
extract_features(os.path.join("data", "GOPR2713.MP4"), os.path.join("data", "extracted"), 10)
```

```python
def crop_images(extracted_images_path):
    for file in os.listdir(extracted_images_path):
        if file.endswith(".jpg"):
            image = Image.open(os.path.join(extracted_images_path, file))
            image = image.crop((0,0,1564,880))
            image.save(os.path.join(extracted_images_path, file))
```

Permet de rogner le tableau de bord qui était visible afin de ne garder que la route.

```python
crop_images(os.path.join("data", "extracted"))
```
